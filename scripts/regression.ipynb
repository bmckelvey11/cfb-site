```python
from db import get_table
import pandas as pd
import seaborn as sns

games = get_table("core.v_game_book_median",read_only=True)
games.columns
```

```python
# drop rows with missing median_total_open
games = games.dropna(subset=['median_total_open'])
# calculate total points scored in the game
games['total_actual'] = games['home_points'] + games['away_points']

residuals = games['total_actual'] - games['median_total_open']

sns.residplot(x='median_total_open', y='total_actual', data=games, lowess=True, line_kws={'color': 'red'})
```
